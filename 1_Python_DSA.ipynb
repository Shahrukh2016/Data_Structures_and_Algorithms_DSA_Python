```python
### Extraction of digits
# Checking the number is palandrome or not
n = 1221
num = n
result = 0
while num > 0:
  id = num % 10
  result = (result*10)+id
  num = num//10

if result == n:
  print(f"result: {result}, n: {n}")
  print("palindrome")
else:
  print(f"result: {result}, n: {n}")
  print("not palandrome")
```

```python
### Extraction of digits
# Checking the number is armstrong
def isArmstrong(n):
  num = n
  result = 0
  nod = len(str(num))
  while num > 0:
    last_num = num%10
    result = last_num**nod + result
    num = num//10
  return result == n

## Test:
print(isArmstrong(n = 1634))
print(isArmstrong(n = 46))
```

```python
### Extraction of digits
## Print All Factors of a Given Number
## My way
def primeFactors(n):
  result = []
  num = n
  for i in range(1, num+1):
    if num%i==0:
      result.append(i)
  return result

primeFactors(10)

##########################################################

## Optimized way
def primeFactors(n):
  result = []
  num = n
  for i in range(1, num//2+1):
    if num%i==0:
      result.append(i)
  result.append(num)
  return result

primeFactors(10)
```
