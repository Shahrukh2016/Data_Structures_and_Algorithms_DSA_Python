```python
### Extraction of digits
# Checking the number is palandrome or not
n = 1221
num = n
result = 0
while num > 0:
  id = num % 10
  result = (result*10)+id
  num = num//10

if result == n:
  print(f"result: {result}, n: {n}")
  print("palindrome")
else:
  print(f"result: {result}, n: {n}")
  print("not palandrome")
```

```python
### Extraction of digits
# Checking the number is armstrong
def isArmstrong(n):
  num = n
  result = 0
  nod = len(str(num))
  while num > 0:
    last_num = num%10
    result = last_num**nod + result
    num = num//10
  return result == n

## Test:
print(isArmstrong(n = 1634))
print(isArmstrong(n = 46))
```

#### 2. Integer and character Hashing:

```python
#### Finding dictionary frequency
n = [1,2,3,1,2,2,2,2,5,6,8]
num = n
dic = {}
for i in num:
  if i in dic.keys():
    dic[i] = dic[i] + 1
  else:
    dic[i] = i
dic


#### Other way using get method
lst = [1,2,3,1,2,2,2,2,5,6,8]
num = lst
n = len(lst)
dic = {}
for i in range(0,n):
  dic[num[i]] = dic.get(num[i], 0) + 1
dic
```

```python
### Character Hashing (Constraints: 'a' <= s[i] <= 'z')
def FindingOccurence(n, m):
  dic = {}
  hash_list = [0] * 26

  for ch in n:
    ascii_value = ord(ch)
    index = ascii_value - 97
    hash_list[index] += 1

  for ch in m:
    ascii_value = ord(ch)
    index = ascii_value - 97
    dic[ch] = hash_list[index]

  return dic

n = 'sdthyrdjgestsgfaawegehrnynjutjscscaxesd'
m = ['a', 'd', 'z']
print(FindingOccurence(n,m))
```

```python
### Character Hashing when no containts given
def FindingOccurence(n, m):
  n = [z for z in n]
  dic = {}
  for i in n:
    if i in dic.keys():
      dic[i] += 1
    else:
      dic[i] = 1

  res = {}
  for j in m:
    if j in dic.keys():
      res[j] = dic[j]
    else:
      res[j] = 0

  return res

n = 'sdthyrdjgestsgfaawegehrnynjutjscscaxesd'
m = ['a', 'd', 'z']
print(FindingOccurence(n,m))
```
